$$\text{bits por muestra} = \texttt{arreglo.dtype.itemsize} \times 8$$

$$\text{memoria}_{KB} = \frac{N \times \text{bits por muestra}}{8 \times 1024}$$

## Fórmulas que intervienen

**Paso de cuantización.** Con $b$ bits hay $2^b$ niveles, y el ancho de cada
escalón de intensidad es:

$$\Delta = \frac{256}{2^{b}}$$

**RMSE (Root Mean Square Error).** Mide la distancia promedio entre la imagen
original y la cuantizada:

$$\text{RMSE} = \sqrt{\frac{1}{N}\sum_{i=1}^{N}\left(x_i - \hat{x}_i\right)^2}$$

donde $x_i$ es el píxel original, $\hat{x}_i$ el cuantizado y $N$ el total de
píxeles. Se eleva al cuadrado para que los errores positivos y negativos no se
cancelen, y se saca raíz para volver a las unidades originales (niveles de gris).

**RMSE teórico.** El cuantizador que usamos trunca hacia abajo (división entera),
por lo que el error está distribuido uniformemente en $[0, \Delta)$. Se puede
demostrar que:

$$\text{RMSE}_{\text{teórico}} = \frac{\Delta}{\sqrt{3}} \approx 0.577\,\Delta$$

Usted comparará su RMSE medido contra este valor teórico.

## Fórmulas que intervienen

**Frecuencia de Nyquist.** Es el techo de lo representable con una frecuencia de
muestreo $f_s$:

$$f_N = \frac{f_s}{2}$$

**Teorema de Nyquist-Shannon.** Para capturar correctamente una señal cuya
frecuencia máxima es $f_{\max}$, se requiere:

$$f_s > 2 \, f_{\max}$$

**Frecuencia del alias.** Si una componente de frecuencia $f$ supera $f_N$, no
desaparece: se refleja y se percibe como otra frecuencia distinta:

$$f_{\text{alias}} = \left| f - f_s \cdot \text{round}\!\left(\frac{f}{f_s}\right) \right|$$
